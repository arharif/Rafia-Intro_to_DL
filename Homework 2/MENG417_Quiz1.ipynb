# Question 1 — Linear Separability, Bias, and Perceptron Design

Let the two classes be represented by the points above.

## Question 1(a) — Is the Data Linearly Separable?

Yes.

A set of two classes is linearly separable if there exists a line

$$
w_1x_1+w_2x_2+b=0
$$

such that every point of one class lies on one side and every point of the other class lies on the opposite side.

Look only at the first coordinate.

For Class 1,

$$
x_1\in\{1,2,1\},
$$

so

$$
\max_{\mathbf x\in\mathcal C_1}x_1=2.
$$

For Class 2,

$$
x_1\in\{3,3,4,4\},
$$

so

$$
\min_{\mathbf x\in\mathcal C_2}x_1=3.
$$

Therefore there is an empty gap

$$
2<x_1<3.
$$

Any vertical line

$$
x_1=c,
\qquad
2<c<3,
$$

separates the two classes.

The midpoint is

$$
c=\frac{2+3}{2}=2.5.
$$

Hence one convenient separator is

$$
\boxed{x_1=2.5}.
$$

### Final answer

$$
\boxed{
\text{Yes, the data are linearly separable.}
}
$$

## Question 1(b) — Do We Need a Bias Term?

Yes, a bias term is required for a correct single-perceptron separator.

### Why?

Without a bias, a perceptron boundary has the form

$$
w_1x_1+w_2x_2=0.
$$

Every such line passes through the origin.

Now compare these two professor-provided points:

$$
\begin{bmatrix}1\\1\end{bmatrix}
\in\mathcal C_1
$$

and

$$
\begin{bmatrix}3\\3\end{bmatrix}
\in\mathcal C_2.
$$

They lie on the same ray from the origin because

$$
\begin{bmatrix}3\\3\end{bmatrix}
=
3
\begin{bmatrix}1\\1\end{bmatrix}.
$$

For any weight vector $\mathbf w$,

$$
\mathbf w^T
\begin{bmatrix}3\\3\end{bmatrix}
=
3
\mathbf w^T
\begin{bmatrix}1\\1\end{bmatrix}.
$$

Therefore the two scores always have the same sign unless both are exactly zero.

So a bias-free homogeneous separator cannot assign these two positive scalar multiples to opposite classes.

Thus,

$$
\boxed{
\text{Yes, a nonzero bias is needed.}
}
$$

## Question 1(c) — Construct a Perceptron by Hand

Use the separator obtained above:

$$
\mathbf w=
\begin{bmatrix}
1\\
0
\end{bmatrix},
\qquad
b=-2.5.
$$

The net input is

$$
a
=
x_1-2.5.
$$

Use the signum activation

$$
\operatorname{sgn}(a)
=
\begin{cases}
+1,&a>0,\\
-1,&a<0.
\end{cases}
$$

There are no training points exactly on the chosen boundary, so the value at $a=0$ does not affect the training classification.

Our class convention is

$$
-1\Rightarrow\mathcal C_1,
$$

$$
+1\Rightarrow\mathcal C_2.
$$

## Verify every point by hand

### Class 1

For

$$
(1,1):
$$

$$
a=1-2.5=-1.5<0
$$

so

$$
\operatorname{sgn}(a)=-1
\Rightarrow
\mathcal C_1.
$$

For

$$
(2,1):
$$

$$
a=2-2.5=-0.5<0
$$

so it is Class 1.

For

$$
(1,2):
$$

$$
a=1-2.5=-1.5<0
$$

so it is Class 1.

### Class 2

For either point with $x_1=3$,

$$
a=3-2.5=0.5>0,
$$

so the output is $+1$.

For either point with $x_1=4$,

$$
a=4-2.5=1.5>0,
$$

so the output is also $+1$.

Therefore,

$$
\boxed{
\mathbf w=
[1,0]^T,

is a correct perceptron for the professor's data.

## Note on uniqueness

This is not the only correct perceptron.

For example,

$$
x_2=2.5
$$

and several diagonal lines also separate the classes.

A linearly separable dataset normally has many valid separating hyperplanes.

```python
# Verify the hand-designed perceptron.

w_hand = np.array([1.0, 0.0])
b_hand = -2.5

X_all = np.vstack([class1, class2])
y_all = np.array([-1, -1, -1, 1, 1, 1, 1])

scores_hand = X_all @ w_hand + b_hand
pred_hand = np.where(scores_hand > 0, 1, -1)

print("Point        Target   Score   Prediction")
print("-"*46)

for point, target, score, pred in zip(X_all, y_all, scores_hand, pred_hand):
    print(f"{str(tuple(point)):14s} {target:6d} {score:7.2f} {pred:11d}")

print("\nAll correct =", np.array_equal(pred_hand, y_all))
```

# Question 2 — Python Code for Question 1

The professor allows the perceptron code and its output to be uploaded as a PDF.

Below are two useful pieces:

1. a direct implementation of the hand-designed classifier;
2. a genuine perceptron-learning loop starting from zero weights.

The first reproduces the exact hand answer.

The second demonstrates the perceptron learning algorithm and shows that it may converge to a **different but equally valid** separating line.

```python
# Question 2A — Direct implementation of the hand-designed perceptron

def signum(a):
    if a > 0:
        return 1
    elif a < 0:
        return -1
    else:
        return 0

w = np.array([1.0, 0.0])
b = -2.5

print("Hand-designed perceptron")
print("w =", w)
print("b =", b)
print()

for x, target in zip(X_all, y_all):
    a = w @ x + b
    y_hat = signum(a)

    print(
        "x =", x,
        " target =", target,
        " net =", a,
        " output =", y_hat
    )
```

## Optional learning verification

For a classical perceptron-learning demonstration, initialize

$$
\mathbf w^{(0)}=
\begin{bmatrix}
0\\
0
\end{bmatrix},
\qquad
b^{(0)}=0,
$$

use learning rate

$$
\eta=1,
$$

and update when

$$
y_i(\mathbf w^T\mathbf x_i+b)\le0.
$$

The update is

$$
\boxed{
\mathbf w
\leftarrow
\mathbf w+\eta y_i\mathbf x_i
}
$$

and

$$
\boxed{
b
\leftarrow
b+\eta y_i.
}
$$

Because the classes are linearly separable, the perceptron convergence theorem guarantees convergence after finitely many mistakes.

The exact final weights depend on initialization and sample order, so they do not need to equal the hand-designed weights.

```python
# Question 2B — Perceptron learning from zero initialization

w_train = np.zeros(2)
b_train = 0.0
eta = 1.0

history = []

for epoch in range(100):

    mistakes = 0

    for i, (x, target) in enumerate(zip(X_all, y_all), start=1):

        score = w_train @ x + b_train

        if target * score <= 0:

            history.append({
                "epoch": epoch + 1,
                "sample": i,
                "x": x.copy(),
                "target": int(target),
                "old_w": w_train.copy(),
                "old_b": b_train,
                "old_score": score
            })

            w_train = w_train + eta * target * x
            b_train = b_train + eta * target

            mistakes += 1

    if mistakes == 0:
        break

scores_train = X_all @ w_train + b_train
pred_train = np.where(scores_train > 0, 1, -1)

print("Converged after", epoch + 1, "epoch(s)")
print("Number of updates =", len(history))
print("Learned w =", w_train)
print("Learned b =", b_train)
print("All training points correct =", np.array_equal(pred_train, y_all))
print()

print("Point        Target   Score   Prediction")
print("-"*46)

for point, target, score, pred in zip(X_all, y_all, scores_train, pred_train):
    print(f"{str(tuple(point)):14s} {target:6d} {score:7.2f} {pred:11d}")
```

With the deterministic data order used above, the learning algorithm converges to another valid separator:

$$
\boxed{
\mathbf w=
\begin{bmatrix}
1\\
3
\end{bmatrix},
\qquad
b=-11.
}
$$

Its decision line is

$$
x_1+3x_2-11=0.
$$

It correctly classifies all seven points.

This does **not** contradict the hand solution

$$
x_1-2.5=0.
$$

Both are valid because separating hyperplanes are not unique.

```python
# Plot both valid separators.

xx = np.linspace(-0.5, 4.8, 300)

# Learned line: x1 + 3*x2 - 11 = 0
# x2 = (11-x1)/3
yy_train = (11 - xx)/3

plt.figure(figsize=(7,6))
plt.scatter(class1[:,0], class1[:,1], marker="o", s=100, label="Class 1")
plt.scatter(class2[:,0], class2[:,1], marker="x", s=100, label="Class 2")

plt.axvline(2.5, linewidth=2, label="Hand: $x_1=2.5$")
plt.plot(xx, yy_train, linestyle="--", linewidth=2,
         label="Learned: $x_1+3x_2-11=0$")

plt.xlabel("$x_1$")
plt.ylabel("$x_2$")
plt.title("Question 2 — Two Valid Perceptron Separators")
plt.xlim(-0.5,4.8)
plt.ylim(-0.5,4.8)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

# Question 3 — Supervised or Unsupervised Learning?

The problem in Question 1 is a **supervised learning** problem.

Why?

Each input vector has a known desired class.

For example,

$$
\begin{bmatrix}
1\\
1
\end{bmatrix}
$$

is known to belong to Class 1, while

$$
\begin{bmatrix}
3\\
3
\end{bmatrix}
$$

is known to belong to Class 2.

During perceptron learning, the algorithm compares its predicted side of the boundary with the supplied target label.

If the prediction is wrong, the known target tells the algorithm how to update the parameters.

Therefore,

$$
\boxed{
\text{Question 1 is supervised learning.}
}
$$

It is not unsupervised because the correct classes are already provided.

# MENG 417 Quiz #1 — Mature Cheat Sheet

## 1. Perceptron model

A perceptron computes

$$
\boxed{
a=\mathbf w^T\mathbf x+b.
}
$$

Then an activation function maps $a$ to a class.

---

## 2. Decision boundary

The decision boundary is

$$
\boxed{
\mathbf w^T\mathbf x+b=0.
}
$$

In two dimensions, this is a line.

---

## 3. Linear separability

Two classes are linearly separable if one line can place every point of one class on one side and every point of the other class on the opposite side.

---

## 4. Bias

Without bias,

$$
\mathbf w^T\mathbf x=0
$$

must pass through the origin.

With bias,

$$
\mathbf w^T\mathbf x+b=0
$$

can be shifted.

This extra freedom is often essential.

---

## 5. Quick test for needing bias

If two samples from opposite classes are positive scalar multiples,

$$
\mathbf x_2=c\mathbf x_1,
\qquad
c>0,
$$

then a bias-free classifier cannot give them opposite signs because

$$
\mathbf w^T\mathbf x_2
=
c\,\mathbf w^T\mathbf x_1.
$$

That is exactly what happens with

$$
(1,1)
$$

and

$$
(3,3).
$$

---

## 6. Signum activation

A common convention is

$$
\operatorname{sgn}(a)
=
\begin{cases}
+1,&a>0,\\
-1,&a<0.
\end{cases}
$$

Always state which class corresponds to $+1$ and which to $-1$.

---

## 7. Perceptron learning rule

With targets

$$
y_i\in\{-1,+1\},
$$

a sample is correctly classified when

$$
y_i(\mathbf w^T\mathbf x_i+b)>0.
$$

If

$$
y_i(\mathbf w^T\mathbf x_i+b)\le0,
$$

update:

$$
\boxed{
\mathbf w
\leftarrow
\mathbf w+\eta y_i\mathbf x_i
}
$$

and

$$
\boxed{
b
\leftarrow
b+\eta y_i.
}
$$

---

## 8. Perceptron convergence

For linearly separable data, the classical perceptron algorithm converges after finitely many mistakes.

The exact separator is generally not unique.

Different:

- initial weights;
- data orders;
- learning rates

can produce different valid separating lines.

---

## 9. Supervised learning

Supervised learning uses labeled pairs

$$
(\mathbf x_i,y_i).
$$

The target $y_i$ tells the algorithm whether its prediction was correct.

---

# Common Mistakes to Avoid

1. Do not forget the bias term when the separator should not pass through the origin.
2. Do not confuse a valid separator with the only possible separator.
3. Do not forget to state the class-label convention.
4. Do not classify a point using only its coordinates; first compute the perceptron net input.
5. Do not assume the Python-trained weights must equal the hand-designed weights.
6. When a point lies exactly on a decision boundary during training, state how your algorithm treats it.
7. Always substitute the final weights back into every training point to verify correctness.

---
